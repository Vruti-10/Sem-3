```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    if num % 3 == 0:
        print("Number is divisible by both 2 and 3")
    else:
        print("Number is divisible by 2 but not by 3")
else:
    if num % 3 == 0:
        print("Number is divisible by 3 but not by 2")
    else:
        print("Number is not divisible by 2 or 3")
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a <= b and a <= c:
    print("Smallest =", a)
elif b <= a and b <= c:
    print("Smallest =", b)
else:
    print("Smallest =", c)
```

```python
num = int(input("Enter a number: "))

count = num
square = 0

while count > 0:
    square = square + num
    count = count - 1

print("Square =", square)
```

```python
numbers = [10, 20, 30, 40, 50]

total = 0

for num in numbers:
    total = total + num

print("Sum =", total)
```

```python
numbers = [10, None, 20, 30, None, 40]

total = 0

for num in numbers:
    if num is None:
        continue

    total = total + num

print("Sum =", total)
```

```python
numbers = [10, 20, 30, 40, 50]

target = 30

for num in numbers:
    if num == target:
        print("Target found:", num)
        break

    print("Checking:", num)
```

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("A =", A)
print("B =", B)

print("Union =", A | B)
print("Intersection =", A & B)
print("Difference =", A - B)
print("Symmetric Difference =", A ^ B)
```

```python
A = {1, 2, 3}
B = {2, 3, 4}

A.update(B)

print("After update:", A)

A.intersection_update(B)

print("After intersection_update:", A)

print("A is subset of B:", A.issubset(B))
print("A is superset of B:", A.issuperset(B))
```

```python
numbers = [1, 2, 3, 4, 5, 6]

squares = [x * x for x in numbers if x % 2 == 0]

print("Squares of even numbers:", squares)
```
